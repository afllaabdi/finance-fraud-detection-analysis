```python
missing = df.isnull().sum()

missing[missing > 0]
```

```python
categorical_cols = df.select_dtypes(
    include='object'
).columns

categorical_cols
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in categorical_cols:
    df[col] = le.fit_transform(df[col])
```

```python
target = 'Is_Fraud'
```

```python
X = df.drop(target, axis=1)

y = df[target]
```

```python
corr = df.corr(numeric_only=True)
```

```python
corr_target = corr[target].abs()

corr_target.sort_values(
    ascending=False
)
```

```python
import matplotlib.pyplot as plt

corr_target.sort_values().plot(
    kind='barh',
    figsize=(10,6)
)

plt.title(
    "Feature Correlation with Target"
)

plt.show()
```

```python
selected_corr = corr_target[
    corr_target > 0.05
].index.tolist()

selected_corr.remove(target)
```

```python
from sklearn.linear_model import LogisticRegression
from mlxtend.feature_selection import SequentialFeatureSelector
```

```python
lr = LogisticRegression(
    max_iter=1000
)
```

```python
sfs = SequentialFeatureSelector(
    lr,
    k_features='best',
    forward=False,
    floating=False,
    scoring='accuracy',
    cv=5
)

sfs.fit(X, y)
```

```python
selected_backward = list(
    sfs.k_feature_names_
)

print(selected_backward)
```

```python
X_selected = X[selected_backward]

processed_df = pd.concat(
    [X_selected, y],
    axis=1
)

processed_df.to_csv(
    'processed_dataset.csv',
    index=False
)
```
